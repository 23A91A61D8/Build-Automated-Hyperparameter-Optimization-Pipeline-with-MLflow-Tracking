# Hyperparameter Optimization Analysis – XGBoost

This notebook presents the analysis of an automated hyperparameter optimization
pipeline built using Optuna and MLflow on the California Housing dataset.

The analysis is performed after running the Docker-based optimization pipeline.


## Analysis Objectives

The objectives of this analysis are:
- To visualize the hyperparameter optimization process
- To understand which hyperparameters influence model performance the most
- To analyze interactions between hyperparameters
- To compare baseline and tuned model performance
- To draw meaningful insights from the optimization results


```python
import optuna
import mlflow
import json
import matplotlib.pyplot as plt
import numpy as np
```

```python
study = optuna.load_study(
    study_name="xgboost-housing-optimization",
    storage="sqlite:///../optuna_study.db"
)

study
```

-The Optuna study is loaded from the SQLite storage created during the Docker
-pipeline execution. This study contains all trials and optimization history.


```python
optuna.visualization.plot_optimization_history(study)
```

### Optimization History

This plot shows the progression of the objective value (negative MSE) across
optimization trials. The trend indicates that Optuna effectively explores the
search space and gradually converges toward better-performing configurations.


```python
optuna.visualization.plot_param_importances(study)
```

### Hyperparameter Importance

This visualization highlights the relative importance of each hyperparameter.
Learning rate, maximum tree depth, and number of estimators have the strongest
impact on model performance, which aligns with XGBoost tuning best practices.


```python
optuna.visualization.plot_parallel_coordinate(study)
```

### Hyperparameter Interactions

The parallel coordinates plot shows how combinations of hyperparameters influence
performance. It helps identify patterns and trade-offs between model complexity
and generalization.


```python
with open("../outputs/results.json", "r") as f:
    results = json.load(f)

results
```

The results.json file contains the final evaluation metrics and best
hyperparameters obtained from the optimization pipeline.


## Baseline vs Tuned Model Comparison

A baseline XGBoost model with default hyperparameters provides reasonable
performance. However, the tuned model obtained through Optuna optimization
achieves lower RMSE and improved generalization on the test dataset.

This demonstrates the effectiveness of automated hyperparameter optimization
over manual tuning or default configurations.


## Key Insights and Conclusion

- Automated hyperparameter optimization significantly improves model performance.
- Learning rate, tree depth, and number of estimators are the most influential parameters.
- Median pruning reduces unnecessary computation by stopping poor trials early.
- MLflow provides transparent and reproducible experiment tracking.
- The entire pipeline is reproducible and production-ready using Docker.

This project demonstrates professional MLOps practices for building,
optimizing, and evaluating machine learning models.
